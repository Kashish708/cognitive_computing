```python
def chatbot():
    print("Welcome to HealthBot!")
    print("How can I help you today?")

    print("\nChoose an option:")
    print("1. Enter symptoms")
    print("2. Book appointment")
    print("3. First-aid advice")
    print("4. Exit")

    choice = input("Your choice (1-4): ")

    if choice == '1':

     symptoms = input("Enter your symptoms (comma-separated): ").lower()
     diagnose(symptoms)

    elif choice == '2':
      name = input("Enter your name: ")
      date = input("Enter appointment date (dd-mm-yyyy): ")
      print(f"Appointment booked for {name} on {date}")

    elif choice == '3':
      print(" Basic First-aid Tips:")
      print(" - For cuts: Wash with clean water, apply antiseptic, bandage.")
      print(" - For burns: Cool the burn with water, cover with clean cloth.")
      print(" - For fever: Drink fluids, take rest, use cold compress.")

    elif choice == '4':
      print(" Thank you for using HealthBot. Stay healthy!")


    else:
      print(" Invalid option. Please try again.")



def diagnose(symptoms):
    rules = {
        "fever,cough,sore throat": "You may have the Flu. Stay hydrated and rest.",
        "fever,cough,shortness of breath": "You may have COVID-19. Please consult a doctor.",
        "headache,nausea": "Possible migraine. Avoid bright lights and take rest.",
        "stomach pain,diarrhea": "Could be food poisoning. Drink ORS and rest.",
        "chest pain,shortness of breath": "Critical. Seek emergency medical help!"
    }

    matched = False
    for key in rules:
        if all(sym.strip() in symptoms for sym in key.split(',')):
            print(f"🩺 Diagnosis: {rules[key]}")
            matched = True
            break

    if not matched:
        print(" Sorry, your symptoms do not match any known rules. Please consult a doctor.")



chatbot()
```
